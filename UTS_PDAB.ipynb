# **2. Pembersihan Data**

**A. Pengecekan Missing Values**

Pada tahap ini dilakukan pengecekan terhadap nilai yang hilang pada dataset training dan testing.

```python
train.isnull().sum()
test.isnull().sum()
```

Interpretasi:
semua nilai 0, artinya tidak terdapat missing values pada seluruh variabel.

**B. Pengecekan Data Duplikat**

Dilakukan pengecekan terhadap data duplikat untuk memastikan tidak ada data yang terduplikasi yang dapat menyebabkan bias dalam pelatihan model.

```python
train.duplicated().sum()
```

```python
train = train.drop_duplicates()
```

Berdasarkan hasil pengecekan data duplikat, diperoleh nilai sebesar 0 yang menunjukkan bahwa tidak terdapat data yang terduplikasi dalam dataset training. Dengan demikian, tidak diperlukan proses penghapusan data duplikat karena seluruh data sudah bersifat unik dan dapat digunakan secara langsung dalam proses pelatihan model.

**C. Pemisahan Fitur dan Target**

Pada tahap ini dilakukan pemisahan antara variabel independen (fitur) dan variabel dependen (target).

```python
X_train = train.drop(['quality'], axis=1)
y_train = train['quality']
```

```python
test_id = test['Id']
X_test = test.drop(['Id'], axis=1)
```

Variabel quality digunakan sebagai target karena merupakan nilai yang ingin diprediksi dalam permasalahan klasifikasi ini. Sedangkan variabel lainnya digunakan sebagai fitur untuk membangun model.

**D. Feature Scaling**

Feature scaling dilakukan menggunakan metode StandardScaler untuk menstandarisasi nilai fitur sehingga memiliki rata-rata 0 dan standar deviasi 1.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Drop the 'Id' column from X_train to match the features in X_test
# that was created by dropping 'Id'.
# This ensures both X_train and X_test have the same feature set for scaling.
X_train_for_scaling = X_train.drop('Id', axis=1)

X_train_scaled = scaler.fit_transform(X_train_for_scaling)
X_test_scaled = scaler.transform(X_test)
```

Proses ini penting karena beberapa algoritma machine learning sensitif terhadap skala data. Dengan scaling, model dapat belajar dengan lebih optimal dan konvergen lebih cepat.

# **3. Pembuatan Model**

**a. Import & Training Model**

Model yang digunakan dalam penelitian ini adalah Random Forest Classifier. Model ini dipilih karena mampu menangani data dengan kompleksitas tinggi dan hubungan non-linear antar variabel. Selain itu, Random Forest juga memiliki keunggulan dalam mengurangi overfitting melalui teknik ensemble learning.

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(random_state=42)
model.fit(X_train_scaled, y_train)
```

Random Forest Classifier dipilih karena memiliki kemampuan yang baik dalam menangani data numerik dengan pola hubungan yang kompleks dan non-linear. Selain itu, model ini mampu mengurangi risiko overfitting melalui kombinasi banyak decision tree (ensemble learning).

Random Forest juga relatif stabil terhadap noise dan mampu memberikan performa yang baik meskipun terdapat variasi antar fitur pada dataset.

**b. Evaluasi Model**

Pada tahap ini digunakan model Random Forest Classifier untuk melakukan klasifikasi kualitas wine. Model ini dipilih karena mampu menangani hubungan non-linear antar variabel serta memiliki performa yang stabil melalui pendekatan ensemble learning.

Model dilatih menggunakan dataset training yang telah melalui tahap preprocessing, termasuk pembersihan data dan feature scaling. Proses pelatihan bertujuan untuk mempelajari pola hubungan antara variabel fitur dengan variabel target yaitu quality.

```python
from sklearn.metrics import accuracy_score, confusion_matrix

y_pred_train = model.predict(X_train_scaled)

print("Accuracy:", accuracy_score(y_train, y_pred_train))
print("Confusion Matrix:\n", confusion_matrix(y_train, y_pred_train))
```

Berdasarkan hasil evaluasi, model menghasilkan nilai akurasi sebesar 1.0 (100%) pada data training. Hal ini menunjukkan bahwa model mampu mengklasifikasikan seluruh data training dengan sangat baik tanpa kesalahan.

Hasil confusion matrix menunjukkan bahwa seluruh nilai berada pada diagonal utama dan tidak terdapat nilai di luar diagonal. Hal ini mengindikasikan bahwa semua data berhasil diprediksi dengan benar oleh model, sehingga tidak terjadi kesalahan klasifikasi antar kelas.

Meskipun demikian, nilai akurasi yang sangat tinggi ini perlu diperhatikan karena berpotensi mengindikasikan overfitting, yaitu kondisi di mana model terlalu menyesuaikan diri dengan data training sehingga kurang mampu melakukan generalisasi pada data baru. Oleh karena itu, diperlukan pengujian lebih lanjut menggunakan data testing untuk memastikan bahwa model tetap memiliki performa yang baik.

# **4. Prediksi Data Testing**

**a. Load Model**

```python
import joblib

joblib.dump(model, 'model_wine.pkl')

model_loaded = joblib.load('model_wine.pkl')
```

Model yang telah dilatih kemudian disimpan dalam format .pkl menggunakan library joblib. Proses ini dilakukan agar model dapat digunakan kembali tanpa perlu melakukan pelatihan ulang.

Penyimpanan model merupakan bagian dari deployment sederhana dalam machine learning, karena model dapat diimplementasikan kembali untuk melakukan prediksi pada data baru secara lebih efisien.

**b. Prediksi Data Testing**

```python
y_pred_test = model_loaded.predict(X_test_scaled)
```

Setelah model berhasil dimuat kembali, dilakukan proses prediksi terhadap dataset testing menggunakan model Random Forest yang telah dilatih sebelumnya.

Hasil prediksi berupa nilai quality yang merepresentasikan kualitas wine berdasarkan karakteristik kimia yang dimiliki setiap sampel pada data testing.

Proses ini bertujuan untuk mengetahui kemampuan model dalam melakukan klasifikasi pada data baru yang belum pernah digunakan selama proses pelatihan.

```python
template = pd.read_csv('/content/hasilprediksi_3digitNIMterakhir.csv')

# Isi kolom Quality
template['Quality'] = y_pred_test
```
